```python
class Order:
    def __init__(self,id,date,price,qty):
        self.id=id
        self.date=date
        self.price=price
        self.qty=qty
    def order_info(self):
        print(f"Orderid:{self.id}OrderDate:{self.date}OrderPrice{self.price}OrderQty:{self.qty}")
```

```python
o1=Order(100,"10-10-2025",200,2)
o2=Order(101,"11-12-2025",345,1)
o3=Order(115,"10-01-2026",100,2)
o4=Order(120,"15-02-2026",500,4)
o5=Order(141,"07-07-2026",600,3)
```

```python
o1.order_info()
```

```python
o2.order_info()
```

```python
o3.order_info()
```

```python
from collections import deque
```

```python
order_queue=deque()
```

```python
order_queue.append(o1)
order_queue.append(o2)
order_queue.append(o3)
order_queue.append(o4)
order_queue.append(o5)
```

```python
while order_queue:
    cur=order_queue.popleft()
    print("processing order")
    cur.order_info()
```

```python
class Patient:
    def __init__(self,id,name,dept,type):
        self.id=id
        self.name=name
        self.dept=dept
        self.type=type
    def patient_info(self):
        print(f"Patientid:{self.id}PatientName:{self.name}PatientDept{self.dept}PatientType:{self.type}")    
```

```python
p1=Patient(100,"Hari","cardiac","indoor")
p2=Patient(101,"Amal","psychatry","indoor")
p3=Patient(102,"Nidhi","dermatology","outdoor")
p4=Patient(103,"Thomas","dental","indoor")
p5=Patient(104,"Aliya","ortho","outdoor")
```

```python
p1.patient_info()
```

```python
p3.patient_info()
```

```python
p5.patient_info()
```

```python
import heapq
```

```python
patients=[]
```

```python
heapq.heappush(patients,(3,p1))
heapq.heappush(patients,(2,p2))
heapq.heappush(patients,(5,p4))
heapq.heappush(patients,(1,p3))
heapq.heappush(patients,(4,p5))
```

```python
while patients:
    pr,p=heapq.heappop(patients)
    p.patient_info()
```
