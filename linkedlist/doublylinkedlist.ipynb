```python
class Node:
    def __init__(self,data):
        self.data=data
        self.next=None
        self.prev=None
```

```python
n1=Node(500)
n2=Node(600)
n3=Node(700)
n4=Node(400)
```

```python
n1.next=n2
n2.prev=n1
n2.next=n3
n3.prev=n2
n3.next=n4
n4.prev=n3
```

```python
cur=n1
values=[]
while cur:
    values.append(cur.data)
    cur=cur.next
values.sort()
print(values[-2])
```
