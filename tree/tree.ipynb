```python
class Tree:
    def __init__(self,data):
        self.data=data
        self.left=None
        self.right=None
       
```

```python
root=Tree(5)
root.right=Tree(8)
root.left=Tree(4)
root.left.left=Tree(3)
root.right.right=Tree(100)
```

```python
print(root.left.left.data)
```

```python
print(root.right.right.data)
```

```python
def inorder(root):
        if root==None:
            return 
        inorder(root.left) 
        print(root.data)
        inorder(root.right)
```

```python
inorder(root)
```

```python
def preorder(root):
        if root==None:
            return 
        print(root.data)
        preorder(root.left) 
        preorder(root.right)
```

```python
preorder(root)
```

```python
def postorder(root):
        if root==None:
            return 
        postorder(root.left) 
        postorder(root.right)
        print(root.data)
```

```python
postorder(root)
```

```python
x=int(input("Enter the element:")) 
def search_bst(root,x):
    if root is None:
        print("element not found")     
    if root.data==x:
        print("element found")
        return 
    if x<root.data:
        search_bst(root.left,x)
    else:
        search_bst(root.right,x)
search_bst(root,x)    
```

```python
number=[]
n=int(input("enter the size of list:"))
for i in range(n):
    digit=int(input("enter the elements:"))
    number.append(digit)
number    
```

```python
class Node:
    def __init__(self,data):
        self.data=data
        self.left=None
        self.right=None
      
```

```python
def insert(root,key):
    if root is None:
        return Node(key)
    if key<root.data:
        root.left=insert(root.left,key)
    else:
        root.right=insert(root.right,key)
    return root   
root=None
for i in number:
    root=insert(root,i)
```
