```python
class Student:
    def __init__(self,id,name):
        self.id=id
        self.name=name
    def std_info(self):
        print(f"Id:{self.id} name:{self.name}")
    
```

```python
s1=Student(1,'joe')
```

```python
s1.std_info()
```

```python
s1.id=100
```

```python
s1.std_info()
```

```python
s1.name='don'
```

```python
s1.std_info()
```

```python
class Customer:
    def __init__(self,id,name,address):
        self.id=id
        self.name=name
        self.address=address
    def cust_info(self):
        print(f"id{self.id} name{self.name} address{self.address}")
```

```python
c1=Customer(100,"dawn","america")
```

```python
c1.cust_info()
```

```python
class Cart:
    def __init__(self,name,code,price,qty):
        self.name=name
        self.code=code
        self.price=price
        self.qty=qty
        self.total=price*qty
    @classmethod
    def cart_info(cls):
        print(f"name{self.name} code{self.code} price{self.price} qty{self.qty} total{self.total}")
```

```python
c=Cart("laptop",101,36000,2)
```

```python
c.cart_info()
```
