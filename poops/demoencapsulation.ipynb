```python
class Customer:
    def __init__(self,id,name,address):
        self.id=id
        self.name=name
        self.address=address
    def cust_info(self):
        print(f"id{self.id} name{self.name} address{self.address}")
```

```python
c1=Customer(100,"dawn","america")
```

```python
c1.cust_info()
```

```python
class Cart:
    def  __init__(self,name,code,price,qty):
        self.name=name
        self.code=code
        self.price=price
        self.qty=qty
        self.total=price*qty
    def cart_info(self):
        print(f"name{self.name} code{self.code} price{self.price} qty{self.qty} total{self.total}")
```

```python
c=Cart("laptop",101,36000,2)
```

```python
c.cart_info()
```

```python
class Calculator:
    def __int__(self,a,b):
        self.a=a
        self.b=b
    def add(self):
        return self.a+self.b
    def sub(self):
        return self.a-self.b
    def mul(self):
        return self.a*self.b
    
```

```python
c= Calculator(10,30)
c.add()
```
